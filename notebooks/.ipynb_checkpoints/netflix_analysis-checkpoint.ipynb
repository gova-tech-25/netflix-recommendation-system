# Content-Based Recommendation System

This recommendation system uses:
- TF-IDF Vectorization
- Cosine Similarity

to recommend similar Netflix titles based on movie descriptions.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

## TF-IDF Vectorization

TF-IDF converts text data into numerical vectors so machine learning algorithms can process movie descriptions.

```python
tfidf = TfidfVectorizer(stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(df['description'])
```

## Cosine Similarity

Cosine similarity measures how similar two movie descriptions are by comparing their vector representations.

```python
similarity = cosine_similarity(tfidf_matrix)
```

```python
indices = pd.Series(df.index, index=df['title']).drop_duplicates()
```

## Recommendation Function

```python
def recommend(title):

    if title not in indices:
        print("Movie not found")
        return

    idx = indices[title]

    similarity_scores = list(enumerate(similarity[idx]))

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    similarity_scores = similarity_scores[1:6]

    print("\nRecommended Titles:\n")

    for movie in similarity_scores:

        movie_data = df.iloc[movie[0]]

        print("Title:", movie_data['title'])
        print("Type:", movie_data['type'])
        print("Genre:", movie_data['listed_in'])
        print("Rating:", movie_data['rating'])
        print("-" * 40)
```

# Testing the Recommendation System

```python
recommend("Stranger Things")
```

```python
recommend("Narcos")
```

```python
def recommend(title):

    title = title.lower()

    matches = df[df['title'].str.lower().str.contains(title, na=False)]

    if matches.empty:
        print("Movie not found")
        return

    idx = matches.index[0]

    similarity_scores = list(enumerate(similarity[idx]))

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    similarity_scores = similarity_scores[1:6]

    print(f"\nRecommendations based on: {df.iloc[idx]['title']}\n")

    for movie in similarity_scores:
        print(df.iloc[movie[0]]['title'])
```

```python
recommend("money")
```

```python
recommend("money heist")
```
